```python
df["Total_Marks"] = (
    df["Python"] +
    df["AI"] +
    df["Mathematics"]
)
df[["Student_ID", "Name", "Python", "AI", "Mathematics", "Total_Marks"]]
```

```python
highest_total = np.max(df["Total_Marks"])
print("Highest Total Marks:", highest_total)
```

```python
lowest_total = np.min(df["Total_Marks"])
print("Lowest Total Marks:", lowest_total)
```

```python
class_average = np.mean(df["Total_Marks"])
print("Class Average Total Marks:", round(class_average, 2))
```

```python
df["Average_Marks"] = df["Total_Marks"] / 3
df[["Student_ID", "Name", "Total_Marks", "Average_Marks"]]
```

```python
df["Average_Marks"] = df["Average_Marks"].round(2)
df[["Student_ID", "Name", "Average_Marks"]]
```

```python
def assign_grade(average):
    if average >= 90:
        return "A+"
    elif average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"
```

```python
grades = []
for average in df["Average_Marks"]:
    grade = assign_grade(average)
    grades.append(grade)
df["Grade"] = grades
df[["Student_ID", "Name", "Average_Marks", "Grade"]]
```

```python
print("Number of students:", len(df))
print("Number of grades:", len(grades))
```

```python
def check_pass_fail(average):
    if average >= 40:
        return "PASS"
    else:
        return "FAIL"
```

```python
results = []
for average in df["Average_Marks"]:
    result = check_pass_fail(average)
    results.append(result)
df["Result"] = results
df[["Student_ID", "Name", "Average_Marks", "Grade", "Result"]]
```

```python
pass_count = (df["Result"] == "PASS").sum()
fail_count = (df["Result"] == "FAIL").sum()
print("Number of students passed:", pass_count)
print("Number of students failed:", fail_count)
```

```python
highest_index = df["Average_Marks"].idxmax()
highest_student = df.loc[highest_index]
print("Highest Performing Student")
print("--------------------------")
print("Student ID:", highest_student["Student_ID"])
print("Name:", highest_student["Name"])
print("Average Marks:", highest_student["Average_Marks"])
print("Grade:", highest_student["Grade"])
```

```python
lowest_index = df["Average_Marks"].idxmin()
lowest_student = df.loc[lowest_index]
print("Lowest Performing Student")
print("-------------------------")
print("Student ID:", lowest_student["Student_ID"])
print("Name:", lowest_student["Name"])
print("Average Marks:", lowest_student["Average_Marks"])
print("Grade:", lowest_student["Grade"])
```

```python
overall_average = np.mean(df["Average_Marks"])
print("Overall Class Average:", round(overall_average, 2))
```

```python
top_students = df.sort_values(
    by="Average_Marks",
    ascending=False
).head(5)
print("Top 5 Performing Students")
print("-------------------------")
print(
    top_students[
        ["Student_ID", "Name", "Average_Marks", "Grade"]
    ].to_string(index=False)
)
```

```python
python_average = np.mean(df["Python"])
ai_average = np.mean(df["AI"])
math_average = np.mean(df["Mathematics"])
print("Subject-wise Average Marks")
print("--------------------------")
print("Python      :", round(python_average, 2))
print("AI          :", round(ai_average, 2))
print("Mathematics :", round(math_average, 2))
```

```python
subject_summary = pd.DataFrame({
    "Subject": ["Python", "AI", "Mathematics"],
    "Average_Marks": [
        python_average,
        ai_average,
        math_average
    ]
})
subject_summary["Average_Marks"] = subject_summary["Average_Marks"].round(2)
subject_summary
```

```python
best_subject_index = subject_summary["Average_Marks"].idxmax()
weakest_subject_index = subject_summary["Average_Marks"].idxmin()
best_subject = subject_summary.loc[best_subject_index]
weakest_subject = subject_summary.loc[weakest_subject_index]
print("Best Subject:")
print(
    best_subject["Subject"],
    "-",
    best_subject["Average_Marks"]
)
print("\nWeakest Subject:")
print(
    weakest_subject["Subject"],
    "-",
    weakest_subject["Average_Marks"]
)
```
